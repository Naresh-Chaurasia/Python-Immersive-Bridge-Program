## 2. Simple Prompt Template
Use placeholders like `{city}` so you can reuse the same prompt with different inputs.

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template('Suggest one weekend activity in {city}.')

print(prompt)
```

```python
chain = prompt | llm

result = chain.invoke({'city': 'Pune'})
print(result.content)
```

## 3. Simple Few-Shot Prompt
Give a couple of examples so the model follows your exact output style.

```python
from langchain_core.prompts import FewShotChatMessagePromptTemplate

examples = [
    {'word': 'happy', 'opposite': 'sad'},
    {'word': 'hot', 'opposite': 'cold'},
]

example_prompt = ChatPromptTemplate.from_messages([
    ('human', '{word}'),
    ('ai', '{opposite}')
])

few_shot = FewShotChatMessagePromptTemplate(
    example_prompt=example_prompt,
    examples=examples,
)

prompt = ChatPromptTemplate.from_messages([
    ('system', 'Reply with only the single opposite word.'),
    few_shot,
    ('human', '{word}')
])

chain = prompt | llm
print(chain.invoke({'word': 'summer'}).content)
```
