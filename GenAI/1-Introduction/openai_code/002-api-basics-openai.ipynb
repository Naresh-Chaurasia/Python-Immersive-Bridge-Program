# NB-01 — API Basics

**Learning goal:** Make your first call to the Anthropic API. Understand the request structure, how to read the response, how tokens work, and how streaming changes the UX.

**Concepts covered**
- `anthropic.Anthropic()` client initialisation
- `client.messages.create()` — the core API call
- `model`, `max_tokens`, `messages` parameters
- Reading `response.content[0].text`
- Token counts — `response.usage.input_tokens` and `response.usage.output_tokens`
- Streaming with `client.messages.stream()`

## Cell 1 — Setup

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(
    "/Users/nareshchaurasia/nc/PYTHON-ARCHITECT/Python-Immersive-AI-MAC/.env"
)

# api_key = os.getenv("CO_API_KEY")
# print(api_key)

load_dotenv(dotenv_path=os.path.join("..", ".env"))

assert os.environ.get("OPENAI_API_KEY"), (
    "OPENAI_API_KEY not found. Add it to the project-root .env file."
)

# The client reads OPENAI_API_KEY from the environment automatically.
client = OpenAI()

MODEL = "gpt-6-luna"  # strong general/coding model; or "gpt-6-luna" for fast, high-volume tasks
print(f"Client ready. Using model: {MODEL}")
```

## Cell 2 — First API Call

```python
response = client.responses.create(
    model=MODEL,
    max_output_tokens=256,
    input="What is ICH E2A in pharmaceutical regulation?",
)

print(response.output_text)
```

## Cell 3 — Inspect Token Usage

```python
print(f"Input tokens  : {response.usage.input_tokens}")
print(f"Output tokens : {response.usage.output_tokens}")
print(f"Total tokens  : {response.usage.input_tokens + response.usage.output_tokens}")
```

## Cell 4 — System Message

```python
response = client.responses.create(
    model=MODEL,
    max_output_tokens=256,
    instructions="You are a GenAI research assistant. Answer the following question in a concise manner.",
    input=[
        {"role": "user", "content": "Can you explain to me what is Agentic AI"}
    ],
)

print(response.output_text)
```

## Cell 5 — Streaming

```python
with client.responses.stream(
    model=MODEL,
    max_output_tokens=256,
    input=[
        {
            "role": "user",
            "content": "Can you explain to me what is Agentic AI",
        }
    ],
) as stream:
    for event in stream:
        if event.type == "response.output_text.delta":
            print(event.delta, end="", flush=True)
```

### Key Takeaway

The API call is simple: a `system` message sets the context, a `user` message contains the query, and `max_tokens` controls how much the model can say back. Token counts are your cost meter — log them.
