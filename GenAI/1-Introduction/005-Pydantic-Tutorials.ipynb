# Pydantic Tutorial

This notebook teaches the basics of Pydantic with short explanations followed by runnable examples.

Pydantic is a Python library for data validation and parsing built on type hints. You define a model as a class with typed fields, and Pydantic checks and converts incoming data to match.

## Example 1: Creating a Simple Model

A Pydantic model is created by extending `BaseModel`. Pydantic automatically validates the data types and can even convert compatible values.

```python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int

user = User(name="John", age="25")

print(user)
```

## Example 2: Restricting Values with Literal

`Literal` allows a field to accept only specific values. This is useful when a field should be limited to a predefined list of options.

```python
from typing import Literal
from pydantic import BaseModel

class Ticket(BaseModel):
    priority: Literal["Low", "Medium", "High"]

ticket = Ticket(priority="High")

print(ticket)
```

```python
# This will not work
ticket = Ticket(priority="High")
```

## Example 3: Building a Real-World Model

This example creates a regulatory intake record. The model contains multiple fields, each with validation rules using `Literal`.

## Example 4: Creating a Valid Object

A dictionary containing valid data is passed to the model. Pydantic validates the input and creates a strongly typed object.

```python
from typing import Literal
from pydantic import BaseModel, field_validator


# Define a schema for an online bookstore order
class BookOrder(BaseModel):

    # Restrict format to allowed options
    format: Literal["hardcover", "paperback", "ebook"]

    # Restrict shipping speed to allowed options
    shipping: Literal["standard", "express"]

    # Customer name to validate
    customer_name: str

    # Ensure customer_name is not blank or spaces only
    @field_validator("customer_name")
    @classmethod
    def name_not_empty(cls, v):
        if not v.strip():
            raise ValueError("customer_name cannot be empty")
        return v.strip().title()  # Clean spaces and format nicely

    # Ensure customer_name has at least a first and last name
    @field_validator("customer_name")
    @classmethod
    def require_full_name(cls, v):
        if len(v.split()) < 2:
            raise ValueError("customer_name must include first and last name")
        return v
```

```python



# --- Example Usage ---

# Valid order
order1 = BookOrder(
    format="paperback",
    shipping="express",
    customer_name="maya patel"
)
print(order1)  # customer_name becomes "Maya Patel"
```

## Summary

* `BaseModel` defines the structure of data.
* Type hints provide automatic validation.
* `Literal` restricts values.
* Validators implement custom business rules.
* `ValidationError` handles invalid data.
* `model_dump()` converts the model to a dictionary.
