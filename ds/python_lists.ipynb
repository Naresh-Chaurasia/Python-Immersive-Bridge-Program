# Introduction to Python Lists

A **list** in Python is an ordered, **mutable** (changeable) collection of items. Lists are written with square brackets `[]` and can store elements of any data type, including numbers, strings, and even other lists.

### Key Characteristics of Lists
* **Ordered:** Items maintain a defined order, and new items are added to the end by default.
* **Mutable:** You can add, remove, or modify elements after the list is created.
* **Allow Duplicates:** Lists can contain multiple items with the same value.

### Basic Syntax
```python
my_list = ["apple", "banana", "cherry"]
```

---
## Example 1: Creating and Accessing Lists
You can access list items using **indexing** (starting at `0`) or extract a sub-list using **slicing** (`[start:stop]`).

````python
# Creating a list of fruits
fruits = ["apple", "banana", "cherry", "mango", "orange"]

# Accessing elements using positive and negative indexing
print("First fruit:", fruits[0])        # Index 0 is the first item
print("Last fruit:", fruits[-1])        # Index -1 is the last item

# Slicing: getting a subset of the list (from index 1 up to, but not including, 4)
print("Middle fruits:", fruits[1:4])
````

---
## Example 2: Modifying Lists (Adding, Updating, and Removing)
Because lists are **mutable**, you can change items by index and use built-in methods like `append()`, `insert()`, and `remove()`.

````python
# Initialize a to-do list
tasks = ["Buy groceries", "Clean room", "Pay bills"]


print("Initial tasks:", tasks)
print(tasks[1])


# Update an existing item by index
tasks[1] = "Deep clean room"
print("Initial tasks:", tasks)
print(tasks[1])
````

````python
# Add an item to the end using append()
tasks.append("Call mom")

# Insert an item at a specific position using insert()
tasks.insert(0, "Morning workout")
print("Updated tasks:", tasks)

# Remove a specific item by value using remove()
tasks.remove("Pay bills")
print("After removing a task:", tasks)
````

---
## Example 3: Looping and Useful List Operations
You can iterate through a list with a `for` loop and use built-in functions like `len()`, `sum()`, and `sort()`.

````python
# List of exam scores
scores = [88, 92, 75, 95, 84]

# Sort the list in ascending order (modifies the list in place)
scores.sort()
print("Sorted scores:", scores)
````

````python
# Calculate the average score using sum() and len()
average = sum(scores) / len(scores)
print(f"Average score: {average:.1f}")
````

````python
# Loop through each score and check if it is 90 or above
print("High scores (90+):")
for score in scores:
    print(score)
````

---
## Key Takeaways
1. Use square brackets `[]` to create a list.
2. Lists are **mutable**—use `append()`, `insert()`, `remove()`, or `pop()` to modify their contents.
3. Indexing starts at `0`, negative indices count from the end (`-1`), and slicing (`list[start:end]`) extracts a portion of the list.
