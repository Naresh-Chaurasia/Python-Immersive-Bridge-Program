# 5. Introduction to Python Tuples

A **tuple** in Python is an ordered, **immutable** (unchangeable) collection of items. Tuples are written with round brackets `()` and can store elements of different data types.

### Key Characteristics of Tuples
* **Ordered:** Items have a defined order that will not change.
* **Immutable:** Once created, you cannot add, remove, or modify elements.
* **Allow Duplicates:** Tuples can contain items with the same value.

### Basic Syntax
```python
my_tuple = ("apple", "banana", "cherry")
```

---
## Example 1: Creating and Accessing Tuples
You can access tuple items by referring to their **index number** inside square brackets `[]` (starting at `0`).

````python
# Creating a tuple with mixed data types
student_info = ("Alice", 21, "Computer Science")

# Accessing elements using positive and negative indexing
print("Name:", student_info[0])         # First item
print("Age:", student_info[1])          # Second item
print("Major:", student_info[-1])       # Last item using negative index


# student_info[0] = "Bob"  # This will raise an error because tuples are immutable
````

````python
# Note: A single-item tuple requires a trailing comma
single_item = ("Python",)
print("Type of single_item:", type(single_item))
````

---
## Example 2: Immutability and Tuple Unpacking
Because tuples are **immutable**, trying to change a value raises a `TypeError`. However, you can easily **unpack** tuple values into individual variables.

````python
# Coordinates stored as a tuple
coordinates = (10.5, 20.8)

# Tuple unpacking: assigning each element to a variable
x, y = coordinates
print(f"X coordinate: {x}, Y coordinate: {y}")

# Demonstrating immutability (uncommenting the line below will cause an error)
# coordinates[0] = 15.0  # TypeError: 'tuple' object does not support item assignment

# To "modify" a tuple, you can convert it to a list, change it, and convert back
temp_list = list(coordinates)
temp_list[0] = 15.0
updated_coordinates = tuple(temp_list)
print("Updated coordinates:", updated_coordinates)
````

---
## Example 3: Useful Tuple Methods and Operations
Tuples have two built-in methods: `count()` and `index()`, and support standard operations like concatenation and length checking.

````python
# Tuple with duplicate numbers
numbers = (4, 2, 7, 2, 9, 2, 5)

# Count how many times '2' appears in the tuple
count_twos = numbers.count(2)
print("Number of 2s:", count_twos)

# Find the index of the first occurrence of '7'
index_seven = numbers.index(7)
print("Index of 7:", index_seven)

# Combining (concatenating) two tuples
more_numbers = (10, 12)
combined = numbers + more_numbers
print("Combined tuple:", combined)
print("Total length:", len(combined))
````

---
## Key Takeaways
1. Use parentheses `()` to create a tuple, and include a trailing comma `(item,)` for a single-element tuple.
2. Tuples are **immutable**, making them faster than lists and ideal for fixed data (like coordinates or configuration constants).
3. Use **tuple unpacking** (`a, b = my_tuple`) to cleanly extract values into variables.
