# 6. Introduction to Python Sets

A **set** in Python is an unordered collection of **unique** items. Sets are written with curly braces `{}` and are commonly used to remove duplicates and perform mathematical set operations.

### Key Characteristics of Sets
* **Unordered:** Items do not have a defined index or order.
* **No Duplicates:** Duplicate values are automatically ignored.
* **Mutable:** You can add or remove items after the set is created (though individual elements must be immutable).

### Basic Syntax
```python
my_set = {"apple", "banana", "cherry"}
```

---
## Example 1: Creating Sets and Removing Duplicates
Passing a list with repeated items into the `set()` function automatically filters out duplicates.

````python
# Creating a set directly (duplicates are automatically removed)
colors = {"red", "green", "blue", "red", "yellow", "blue"}
print("Unique colors:", colors)
print(type(colors))  # Output: <class 'set'>
````

````python

# Converting a list with duplicates into a set
visitor_ids = [101, 102, 103, 101, 104, 102]
print(type(visitor_ids))  # Output: <class 'list'>
unique_visitors = set(visitor_ids)
print(type(unique_visitors))  # Output: <class 'set'>
print("Unique visitor IDs:", unique_visitors)
````

````python
# Note: To create an empty set, use set(), NOT {} (which creates an empty dictionary)
empty_set = set()
print("Type of empty_set:", type(empty_set))
````

---
## Example 2: Adding and Removing Elements
You can modify a set using methods like `add()`, `remove()`, and `discard()`.

````python
# Initialize a set of programming languages
languages = {"Python", "Java", "C++"}

# Add a single element using add()
languages.add("Java")
print("After adding JavaScript:", languages)
````

````python
# Remove an element using discard() (does NOT raise an error if item is missing)
languages.discard("Java")
languages.discard("Ruby")  # Safe even though 'Ruby' is not in the set
print("After discarding Java:", languages)
````

````python
isPresent = "Python1" in languages
print("Is Python in the set?", isPresent)
````

---
## Example 3: Mathematical Set Operations
Sets make it easy to compute **unions**, **intersections**, and **differences** between two collections.

````python
# Students enrolled in two different courses
math_students = {"Alice", "Bob", "Charlie", "David"}
science_students = {"Charlie", "David", "Eve", "Frank"}

# Union: all unique students across both courses
all_students = math_students | science_students  # or math_students.union(science_students)
print("All students (Union):", all_students)
````

````python
# Intersection: students taking BOTH courses
both_courses = math_students & science_students  # or math_students.intersection(science_students)
print("Both courses (Intersection):", both_courses)
````

````python
# Difference: students taking Math ONLY
math_only = math_students - science_students     # or math_students.difference(science_students)
print("Math only (Difference):", math_only)
````

---
## Key Takeaways
1. Sets store **unique, unordered** elements and do not support indexing (`my_set[0]` raises a `TypeError`).
2. Use `set()` to create an empty set, as `{}` creates an empty dictionary.
3. Sets are ideal for deduplicating data, fast membership testing (`in`), and operations like union (`|`) and intersection (`&`).
