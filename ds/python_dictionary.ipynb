# 7. Introduction to Python Dictionaries

A **dictionary** in Python is a collection of **key-value pairs**. Instead of accessing items by a numerical index, you look up values using their unique keys. Dictionaries are written with curly braces `{}`.

### Key Characteristics of Dictionaries
* **Key-Value Mapping:** Each key maps to a specific value (`key: value`).
* **Ordered (Python 3.7+):** Dictionaries preserve the insertion order of keys.
* **Mutable & Unique Keys:** You can add, update, or delete pairs, and every key must be unique.

### Basic Syntax
```python
my_dict = {
    "key1": "value1",
    "key2": "value2"
}
```

---
## Example 1: Creating and Accessing a Dictionary
You can retrieve values using square brackets `dict[key]` or the safer `.get(key)` method.

````python
# Creating a dictionary representing a book
book = {
    "title": "The Hobbit",
    "author": "J.R.R. Tolkien",
    "year": 1937
}

# Accessing values using keys
print("Title:", book["title"])
print("Author:", book["author"])

# Using .get() avoids a KeyError if the key does not exist
print("Publisher:", book.get("publisher", "Not specified"))
````

---
## Example 2: Adding, Updating, and Deleting Items
Dictionaries are dynamic—you can easily add new key-value pairs, update existing ones, or remove entries.

````python
# Initialize a user profile dictionary
user = {
    "username": "coder123",
    "level": 1
}

# Add a new key-value pair
user["email"] = "coder123@example.com"

# Update an existing value
user["level"] = 2
print("Updated profile:", user)

# Remove a key-value pair using pop()
removed_email = user.pop("email")
print("Removed email:", removed_email)
print("Final profile:", user)
````

---
## Example 3: Looping Through a Dictionary
You can iterate over keys, values, or both simultaneously using `.keys()`, `.values()`, and `.items()`.

````python
# Dictionary of product prices
prices = {
    "Apple": 1.20,
    "Banana": 0.50,
    "Orange": 0.80
}

# Loop through both keys and values using .items()
print("Store Price List:")
for fruit, price in prices.items():
    print(f"- {fruit}: ${price:.2f}")

# Calculate total cost of all items using .values()
total_cost = sum(prices.values())
print(f"Total cost for one of each: ${total_cost:.2f}")
````

---
## Key Takeaways
1. Dictionaries store data as **`key: value`** pairs inside curly braces `{}`.
2. Use `my_dict.get(key, default)` to safely retrieve values without risking a `KeyError`.
3. Use `.keys()`, `.values()`, and `.items()` to inspect or loop through dictionary contents.
