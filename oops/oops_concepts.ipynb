```python
class Car:
    # Constructor to initialize the object
    def __init__(self):
        print('Inside init')
        print(self)
        print('-----')

    # Constructor to initialize the object
    def __init__(self, brand1, model1):
        print('Inside init...')
        self.brand = brand1
        self.model = model1

    # Method to display car details
    def display(self):
        print('Inside Display')
        print(self.brand)
        print(self.model)
```

```python
c = Car()
print('Print value of c ----- ')
print(c)

c.display()
```

```python
c1 = Car('TATA', 2025)
c1.display()
```
