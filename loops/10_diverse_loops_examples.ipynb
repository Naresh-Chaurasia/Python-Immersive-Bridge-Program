# Advanced Examples of Python `for` and `while` Loops

This notebook provides multiple distinct examples for both `for` loops (iterating across different data types) and `while` loops (handling various conditional scenarios).

---
## Part 1: Diverse `for` Loop Examples
A `for` loop can iterate over many different built-in data structures in Python.

```python
# Example 1: Iterating over a Dictionary (keys, values, and items)
student_grades = {"Alice": 85, "Bob": 92, "Charlie": 78}

print("--- Dictionary Iteration (.items()) ---")
for name, grade in student_grades.items():
    print(f"{name} scored {grade} points.")

print("\n--- Dictionary Iteration (keys & values separately) ---")
for student in student_grades.keys():
    print(f"Student: {student}")
```

```python
# Example 2: Iterating over a Set and a String
# Sets are unordered collections of unique items
unique_tags = {"python", "coding", "beginner", "tutorial"}

print("--- Set Iteration ---")
for tag in sorted(unique_tags):  # Sorting for neat output order
    print(f"Tag: #{tag}")

# Strings are sequences of characters
word = "Python"
print("\n--- String Iteration ---")
for char in word:
    print(f"Character: {char}")
```

```python
# Example 3: Iterating with enumerate() (getting index and value together)
colors = ["red", "green", "blue"]

print("--- Enumerate Iteration ---")
for index, color in enumerate(colors, start=1):
    print(f"Color #{index}: {color}")
```

## Part 2: Diverse `while` Loop Examples
`while` loops run continuously as long as a condition evaluates to `True`.

```python
# Example 1: Password Validation Attempt Counter (User Input simulation)
correct_password = "secret123"
attempts_left = 3

print("--- While Loop: Password Retry Simulation ---")
# Simulating a scenario where a user tries entering a password
while attempts_left > 0:
    # Simulated user attempt (imagine this comes from input())
    guessed_password = "wrong_password" if attempts_left > 1 else "secret123"
    
    if guessed_password == correct_password:
        print("Access granted! Welcome.")
        break
    else:
        attempts_left -= 1
        print(f"Incorrect password. Attempts remaining: {attempts_left}")

if attempts_left == 0:
    print("Account locked due to too many failed attempts.")
```

```python
# Example 2: Exponential Growth / Doubling Condition
# Simulating bacterial growth or compounding steps until a threshold
bacteria_count = 100
max_limit = 1000
hour = 0

print("\n--- While Loop: Growth Threshold Simulation ---")
while bacteria_count < max_limit:
    hour += 1
    bacteria_count *= 2  # Doubles every hour
    print(f"Hour {hour}: Bacteria count reached {bacteria_count}")

print(f"Population exceeded safety limit after {hour} hours.")
```

```python
# Example 3: Interactive Menu / Processing Queue Simulation
print("\n--- While Loop: Processing a Queue ---")
print("Tasks remaining in queue:")
task_queue = ["Task 1", "Task 2", "Task 3"]

# Continue running while there are items in the queue
while len(task_queue) > 0:
    current_task = task_queue.pop(0)  # Remove the first item from the list
    print(f"Processing... [{current_task}] completed.")

print("All tasks processed successfully!")
```

---
## Key Takeaways
1. **`for` loops** work seamlessly with lists, tuples, sets, strings, and dictionaries (via `.items()`, `.keys()`, `.values()`, or `enumerate()`).
2. **`while` loops** are ideal for dynamic conditions such as login attempt limits, threshold tracking (growth/savings), or processing queues until empty.
