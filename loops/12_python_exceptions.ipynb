# Introduction to Python Exceptions (Error Handling)

An **exception** in Python is an error that occurs while a program is running (at runtime), interrupting the normal flow of execution. Using `try` and `except` blocks allows you to catch and handle errors gracefully instead of letting your program crash.

### Why handle exceptions?
* **Prevent Crashes:** Keep your application running smoothly even when unexpected input or missing files occur.
* **User-Friendly Feedback:** Display clear, helpful error messages instead of confusing tracebacks.
* **Safe Cleanup:** Ensure resources (like open files or network connections) are properly closed.

### Basic Syntax
```python
try:
    # Code that might raise an exception
    risky_operation()
except SomeException:
    # Code that runs if that specific exception occurs
    handle_error()
else:
    # Optional: runs if NO exception occurred
    on_success()
finally:
    # Optional: always runs, whether an exception occurred or not
    cleanup()
```

## Example 1: Handling Division by Zero & Invalid Input
When dealing with user input or math calculations, catching specific exceptions like `ZeroDivisionError` and `ValueError` prevents unexpected crashes.

````python
def divide_numbers(a, b):
    """Safely divides two numbers and handles common errors."""
    try:
        # Attempt to convert inputs to floats and divide
        num1 = float(a)
        num2 = float(b)
        result = num1 / num2
    except ZeroDivisionError:
        # Runs if b is 0
        print("Error: You cannot divide by zero!")
    except ValueError:
        # Runs if a or b cannot be converted to a number
        print("Error: Please enter valid numeric values.")
    else:
        # Runs only if the try block succeeded without errors
        print(f"Result: {num1} / {num2} = {result:.2f}")

# Testing different scenarios
divide_numbers(10, 2)       # Valid input
divide_numbers(10, 0)       # Triggers ZeroDivisionError
divide_numbers("abc", 5)    # Triggers ValueError
````

---
## Example 2: Handling Missing Dictionary Keys & List Indexes
Lookups in collections can fail if a key or index does not exist (`KeyError` or `IndexError`).

````python
student_scores = {"Alice": 90, "Bob": 85}

def get_student_score(name):
    """Looks up a student's score safely."""
    try:
        score = student_scores[name]
    except KeyError:
        print(f"Notice: '{name}' was not found in the gradebook.")
    else:
        print(f"{name}'s score is {score}.")

get_student_score("Alice")    # Key exists
get_student_score("Charlie")  # Triggers KeyError
````

## Example 3: Using `finally` and Raising Custom Exceptions (`raise`)
The `finally` block always executes (great for cleanup), and you can intentionally trigger an exception using the `raise` keyword when business rules are violated.

````python
def withdraw_money(balance, amount):
    """Processes a bank withdrawal and validates the amount."""
    try:
        if amount <= 0:
            # Manually raise a ValueError for invalid withdrawal amounts
            raise ValueError("Withdrawal amount must be greater than zero.")
        if amount > balance:
            raise ValueError("Insufficient funds for this withdrawal.")
        
        balance -= amount
        print(f"Withdrawal successful! New balance: ${balance:.2f}")
    except ValueError as err:
        # 'as err' captures the error message we wrote above
        print(f"Transaction Failed: {err}")
    finally:
        # Always runs regardless of success or failure
        print("Thank you for banking with us.\n" + "-" * 35)

withdraw_money(500, 150)   # Valid withdrawal
withdraw_money(500, 700)   # Insufficient funds
withdraw_money(500, -20)   # Invalid negative amount
````

---
## Key Takeaways
1. Place risky code inside a **`try`** block and catch specific errors using **`except ExceptionType:`**.
2. Avoid bare `except:` clauses when possible—catching specific exceptions (like `ValueError`, `KeyError`, `FileNotFoundError`) makes debugging much easier.
3. Use **`else`** for code that should run only when no error happens, and **`finally`** for cleanup steps that must always run.
4. Use **`raise`** to trigger your own exceptions when invalid data or conditions are detected.
