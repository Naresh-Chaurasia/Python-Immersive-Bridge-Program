# Introduction to Python Modules

A **module** in Python is simply a file containing Python definitions, functions, and variables (ending in `.py`) that you can import and reuse in other programs.

### Why use modules?
* **Code Organization:** Break large programs into smaller, manageable files.
* **Reusability:** Write functions once and import them across multiple scripts.
* **Built-in Power:** Access Python's extensive Standard Library (e.g., `math`, `random`, `datetime`).

### Basic Import Syntax
```python
import module_name                 # Import the entire module
from module_name import item     # Import a specific function or variable
import module_name as alias        # Import with a shorter nickname
```

---
## Example 1: Using a Built-in Module (`math`)
Python comes with a large standard library of built-in modules. You load them using the `import` keyword and access their functions using dot notation (`module.item`).

````python
# Import the entire 'math' module
import math

# Use functions and constants from the math module
radius = 5
area = math.pi * math.pow(radius, 2)

print(f"Value of Pi: {math.pi:.4f}")
print(f"Area of circle: {area:.2f}")
print("Square root of 64:", math.sqrt(64))
````

---
## Example 2: Importing Specific Items & Using Aliases (`random` and `datetime`)
Instead of importing an entire module, you can import only what you need using `from ... import ...`, or give a module a shorter nickname using `as`.

````python
# Import only specific functions from the 'random' module
from random import randint, choice

# Roll a 6-sided dice
dice_roll = randint(1, 6)
print("You rolled a:", dice_roll)

# Pick a random winner from a list
participants = ["Alice", "Bob", "Charlie", "Diana"]
winner = choice(participants)
print("Lucky winner:", winner)

# Import 'datetime' with an alias 'dt' for convenience
import datetime as dt

today = dt.date.today()
print("Today's date:", today)
````

---
## Example 3: Creating and Importing Your Own Custom Module
Any `.py` file you write can act as a module for another file in the same folder. The code below creates a `calculator.py` file and then imports it.

````python
# Step 1: Create a custom module file named 'calculator.py'
module_code = '''# File: calculator.py

def add(a, b):
    """Returns the sum of two numbers."""
    return a + b

def multiply(a, b):
    """Returns the product of two numbers."""
    return a * b

TAX_RATE = 0.08
'''

with open("calculator.py", "w") as f:
    f.write(module_code)

# Step 2: Import and use your custom module
import calculator

sum_result = calculator.add(10, 5)
product_result = calculator.multiply(4, 3)
tax = sum_result * calculator.TAX_RATE

print("Sum:", sum_result)
print("Product:", product_result)
print(f"Calculated Tax: ${tax:.2f}")
````

---
## Key Takeaways
1. Use `import module_name` to load an entire module and access items with `module_name.item`.
2. Use `from module_name import item` to import specific functions or variables directly into your namespace.
3. Use `import module_name as alias` to give long module names a convenient shorthand.
4. Any `.py` script you create can be imported as a custom module in another Python file.
