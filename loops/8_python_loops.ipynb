# Introduction to Python Loops (`for` and `while`)

Loops allow you to execute a block of code repeatedly without duplicating code. Python provides two primary loop constructs:
* **`for` loop:** Best for iterating over a known sequence (lists, tuples, dictionaries, sets, strings) or a fixed `range()` of numbers.
* **`while` loop:** Best for repeating an action as long as a given condition remains `True`.

---
## 1. The `for` Loop
A `for` loop steps through each item in a collection or range one by one.

```python
# Example A: Iterating through a list
shopping_list = ["apples", "milk", "bread", "eggs"]

print("My Shopping List:")
for item in shopping_list:
    print(f"- Buy {item}")

print("-" * 30)

# Example B: Using range(start, stop) to repeat a task ('stop' is exclusive)
print("Countdown:")
for i in range(1, 6):
    print(i)

print("Liftoff!")
```

---
## 2. The `while` Loop
A `while` loop keeps running **while** a condition is `True`. Always make sure the condition eventually becomes `False` to prevent an infinite loop.

```python
# Example A: Countdown using a while loop
countdown = 3

while countdown > 0:
    print(countdown)
    countdown -= 1  # Decrease counter by 1 on each iteration

print("Happy New Year!")

print("-" * 30)

# Example B: Saving money until reaching a target goal
savings = 100
target = 500
months = 0

while savings < target:
    savings += 100  # Save $100 each month
    months += 1
    print(f"Month {months}: Saved so far = ${savings}")

print(f"Goal reached in {months} months!")
```

---
## Key Takeaways
1. Use a **`for` loop** when iterating over a sequence or a known number of steps (`range()`).
2. Use a **`while` loop** when looping until a specific condition changes.
3. Always update the condition variable inside a `while` loop so it terminates properly.
