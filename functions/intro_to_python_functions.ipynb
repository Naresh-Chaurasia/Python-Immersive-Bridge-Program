# Introduction to Python Functions

A **function** in Python is a reusable block of code designed to perform a specific task. Instead of writing the same code repeatedly, you can group it into a function and call it whenever you need it.

### Why use functions?
* **Reusability:** Write once, use many times.
* **Modularity:** Break complex problems into smaller, manageable pieces.
* **Readability:** Well-named functions make code easier to understand and maintain.

### Basic Syntax
```python
def function_name(parameters):
    """Optional docstring describing what the function does."""
    # Code block (indented)
    return result  # Optional
```

---
## Example 1: A Simple Greeting Function (No Return Value)
This example demonstrates how to define a basic function that accepts a **parameter** (input) and prints a message.

````python
# Define the function using the 'def' keyword
def greet_user(name):
    """Prints a personalized greeting message."""
    # The 'name' variable inside the function holds the value passed when called
    print(f"Hello, {name}! Welcome to Python functions.")

# Calling (invoking) the function with different arguments
greet_user("Alice")
greet_user("Bob")
greet_user(10)
````

---
## Example 2: Calculating the Area of a Rectangle (Using `return`)
Functions often process inputs and send a result back to the caller using the `return` statement. This allows you to store the output in a variable for later use.

````python
# Define a function that takes two parameters: length and width
def calculate_area(length, width):
    """Calculates and returns the area of a rectangle."""
    area = length * width
    return area  # Sends the calculated value back to where the function was called

# Call the function and store the returned result in a variable
room_area = calculate_area(10, 5)
print("The area of the room is:", room_area)

# You can also use the returned value directly in expressions
total_area = calculate_area(4, 3) + calculate_area(6, 2)
print("Combined area:", total_area)

room_area = calculate_area("hello", 2)
print("The area of the room is:", room_area)
````

## Example 3: Temperature Converter with a Default Parameter
You can assign **default values** to parameters. If the caller does not provide a value for that parameter, Python uses the default.

````python
# 'unit' has a default value of "C" (Celsius)
def convert_temperature(temp=20, unit="C"):
    """
    Converts temperature between Celsius and Fahrenheit.
    - If unit is 'C', converts Celsius to Fahrenheit.
    - If unit is 'F', converts Fahrenheit to Celsius.
    """

    print(f"Converting {temp} degrees with unit '{unit}'.") 

    if unit.upper() == "C":
        converted = (temp * 9/5) + 32
        return f"{temp}°C is equal to {converted:.1f}°F"
    elif unit.upper() == "F":
        converted = (temp - 32) * 5/9
        return f"{temp}°F is equal to {converted:.1f}°C"
    else:
        return "Invalid unit. Please use 'C' or 'F'."

print(convert_temperature())

# Calling without specifying 'unit' (uses default "C")
print(convert_temperature(25))

# Calling with 'unit' explicitly set to "F"
print(convert_temperature(77, unit="F"))
````

---
## Key Takeaways
1. Use `def` to create a function.
2. **Parameters** are variables listed inside the parentheses in the function definition; **arguments** are the actual values passed when calling it.
3. Use `return` to send a result back from a function. Without `return`, a function returns `None` by default.
